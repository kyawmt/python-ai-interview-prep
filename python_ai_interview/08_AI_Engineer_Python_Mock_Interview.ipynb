### Debugging exercise 1: mutable default

**Broken code**


````python
def broken_add(x, items=[]):
    items.append(x)
    return items
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_add(x, items=None):
    if items is None: items=[]
    items.append(x); return items
assert fixed_add(1)==[1] and fixed_add(2)==[2]
````

Default list persists between calls. O(1) amortized per append.


### Debugging exercise 2: dictionary key mismatch

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_lookup(d, user_id):
    return d.get(user_id)
assert fixed_lookup({1:'A'},1)=='A'
````

Do not coerce unless the schema uses string keys. Average O(1).


### Debugging exercise 3: infinite pointers

**Broken code**


````python
def broken_palindrome(s):
    left,right=0,len(s)-1
    while left<right:
        if s[left]!=s[right]: return False
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_palindrome(s):
    left,right=0,len(s)-1
    while left<right:
        if s[left]!=s[right]:return False
        left+=1; right-=1
    return True
assert fixed_palindrome('aba')
````

Pointers never move on matching characters. O(n)/O(1).

### Debugging exercise 5: wrong return

**Broken code**


````python
def broken_fib(n):
    a,b=0,1
    for _ in range(n): a,b=b,a+b
    result=[]
    return result
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_fib(n):
    a,b=0,1
    for _ in range(n):a,b=b,a+b
    return a
assert fixed_fib(6)==8
````

### Debugging exercise 6: modify during iteration

**Broken code**


````python
def broken_remove_even(nums):
    for x in nums:
        if x%2==0: nums.remove(x)
    return nums
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_remove_even(nums):
    return [x for x in nums if x%2]
assert fixed_remove_even([2,2,3])==[3]
````

Removal shifts and skips elements. O(n) solution/output O(n).

### Debugging exercise 8: duplicate loss

**Broken code**


````python
def broken_two_sum(nums,target):
    positions={x:i for i,x in enumerate(nums)}
    for x in nums:
        if target-x in positions:return [positions[x],positions[target-x]]
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
def fixed_two_sum(nums,target):
    seen={}
    for i,x in enumerate(nums):
        if target-x in seen:return [seen[target-x],i]
        seen[x]=i
    return []
assert fixed_two_sum([3,3],6)==[0,1]
````

One-pass ordering prevents using one index twice. O(n)/O(n).


### Debugging exercise 9: NumPy shape mismatch

**Broken code**


````python
def broken_linear(X,W):
    return X*W  # X=(batch,features), W=(features,out)
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
import numpy as np
def fixed_linear(X,W):
    return X @ W
assert fixed_linear(np.ones((2,3)),np.ones((3,4))).shape==(2,4)
````

Need matrix multiplication, not incompatible elementwise broadcasting. O(bfo).


### Debugging exercise 10: pandas precedence

**Broken code**


````python
def broken_filter(df):
    return df[df['age']>30 & df['active']]
````

<details><summary>Hint</summary>Trace a minimal boundary or duplicate case.</details>

#### Fix and explanation


````python
import pandas as pd
def fixed_filter(df):
    return df[(df['age']>30) & df['active']]
assert len(fixed_filter(pd.DataFrame({'age':[31,20],'active':[True,True]})))==1
````

Parenthesize each comparison before `&`. O(n)/O(n) output.

### Round 3 solutions


````python
from collections import defaultdict

def first_repeated_model_id(model_ids):
    seen=set()
    for model_id in model_ids:
        if model_id in seen:return model_id
        seen.add(model_id)
    return None

def longest_confident_run(scores, threshold):
    best=current=0
    for score in scores:
        current=current+1 if score>=threshold else 0
        best=max(best,current)
    return best

def top_average_label(predictions):
    if not predictions:return None
    grouped=defaultdict(list)
    for prediction in predictions:grouped[prediction['label']].append(prediction['score'])
    averages={label:sum(scores)/len(scores) for label,scores in grouped.items()}
    return min(averages, key=lambda label:(-averages[label],label))

assert first_repeated_model_id([1,2,1])==1
assert longest_confident_run([.8,.9,.2,.9],.8)==2
assert top_average_label([{'label':'cat','score':.91},{'label':'dog','score':.82},{'label':'cat','score':.77}])=='cat'
````

**Complexity:** each solution is O(n) time; first/grouping use O(n) or O(k) space, while the run uses O(1). Common mistakes: missing empty behavior, resetting the wrong state, and nondeterministic ties.

### Round 4 solutions


````python
def cosine_similarity(a,b):
    denominator=np.linalg.norm(a)*np.linalg.norm(b)
    return float(np.dot(a,b)/denominator) if denominator else 0.0

def reshape_images(flat, height, width):
    return flat.reshape(flat.shape[0],height,width)

def normalize_rows(X):
    norms=np.linalg.norm(X,axis=1,keepdims=True)
    return X/np.where(norms==0,1,norms)

def prepare_frame(frame, predictions, training_age_median):
    clean=frame.assign(age=frame['age'].fillna(training_age_median))
    clean=clean.loc[clean['score']>=.5]
    summary=clean.groupby('label')['score'].mean()
    joined=clean.merge(predictions,on='user_id',how='left',suffixes=('_observed','_predicted'),validate='many_to_one')
    return summary,joined

assert np.isclose(cosine_similarity(np.array([1.,0.]),np.array([1.,0.])),1)
assert normalize_rows(np.array([[3.,4.]])).shape==(1,2)
````

def f(x):
    r = {}
    for w in x:
        if tuple(sorted(w)) in r:

        else:

<details><summary>Review guidance</summary>Rename, use `defaultdict`, compute key once, and avoid clearing the input.</details>


````python
def group_anagrams_clean(words):
    groups=defaultdict(list)
    for word in words:
        key=tuple(sorted(word))
        groups[key].append(word)
    return list(groups.values())

original=['eat','tea']
assert len(group_anagrams_clean(original))==1 and original==['eat','tea']
````
